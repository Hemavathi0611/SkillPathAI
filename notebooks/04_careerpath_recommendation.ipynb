# CareerPath AI – Career Recommendation System

## Objective
Build a recommendation system that uses the trained CareerPath AI model to suggest suitable career pathways based on a student's education, specialization, academic performance, certifications, and skills.

## Tasks
1. Load the trained machine learning model.
2. Prepare a sample student profile.
3. Generate career predictions.
4. Retrieve the top 5 career recommendations.
5. Display confidence scores.
6. Validate the recommendation output.

## Model Information
- Model: Logistic Regression
- Input features: 8
- Career categories: 30
- Model file: `careerpath_logistic_regression.ipynb.joblib`

## Expected Output
A ranked list of the top 5 predicted career pathways with their model confidence scores.

```python
import pandas as pd
import numpy as np
import joblib

from pathlib import Path
from IPython.display import display

print("Recommendation system libraries imported successfully!")
```

```python
MODEL_PATH = Path("../models/careerpath_logistic_regression.ipynb.joblib")

career_model = joblib.load(MODEL_PATH)

print("CareerPath model loaded successfully!")
print("Model type:", type(career_model).__name__)
print("Number of career categories:", len(career_model.classes_))
```

```python
student_profile = pd.DataFrame([{
    "Education Level": "B.Tech",
    "Specialization": "Computer Applications",
    "Certifications": "Python certification",
    "Score_Type": "CGPA",
    "Score_Value": 8.5,
    "Technical_Skills_Parsed": "Python | SQL | Data analysis | Microsoft Excel",
    "Soft_Skills_Parsed": "Communication | Problem-solving | Teamwork",
    "Skills_Gained_Parsed": "Programming | Analytical thinking | Database management"
}])

display(student_profile)

print("Sample student profile created successfully!")
print("Number of input features:", student_profile.shape[1])
```

```python
# Generate prediction probabilities
probabilities = career_model.predict_proba(student_profile)[0]

# Get predicted career
predicted_career = career_model.predict(student_profile)[0]

print("Predicted Career:")
print(predicted_career)

print("\nPrediction confidence:")
print(f"{probabilities.max() * 100:.2f}%")
```

```python
# Get all career categories and their probabilities
career_probabilities = pd.DataFrame({
    "Career": career_model.classes_,
    "Confidence (%)": probabilities * 100
})

# Sort by confidence in descending order
top_5_recommendations = career_probabilities.sort_values(
    by="Confidence (%)",
    ascending=False
).head(5).reset_index(drop=True)

# Add ranking
top_5_recommendations.index = top_5_recommendations.index + 1
top_5_recommendations.index.name = "Rank"

display(top_5_recommendations)

print("\nTop 5 recommendations generated successfully!")
print("Total probability:", round(probabilities.sum() * 100, 2), "%")
```

```python
# Display all 30 career categories with their confidence scores
all_recommendations = career_probabilities.sort_values(
    by="Confidence (%)",
    ascending=False
).reset_index(drop=True)

all_recommendations.index = all_recommendations.index + 1
all_recommendations.index.name = "Rank"

display(all_recommendations)

print("Total career categories:", len(all_recommendations))
print("Probability total:", round(all_recommendations["Confidence (%)"].sum(), 2), "%")
```

```python
eligibility_mapping = {
    "B.Tech": [
        "M.Tech in Engineering Specialization",
        "M.Tech Computer Science / M.S. in Computer Science",
        "MBA / M.Com / postgraduate specialization in Management"
    ],

    "BCA": [
        "M.Tech Computer Science / M.S. in Computer Science",
        "MBA / M.Com / postgraduate specialization in Management"
    ],

    "B.Com": [
        "M.Com / MBA Finance / professional finance qualification pathway",
        "MBA / M.Com / postgraduate specialization in Management"
    ],

    "BBA": [
        "MBA / M.Com / postgraduate specialization in Management",
        "M.Com / MBA Finance / professional finance qualification pathway"
    ],

    "B.Sc": [
        "M.Sc / postgraduate specialization in a relevant life-science or healthcare field",
        "M.Sc Biotechnology / Microbiology / Life Sciences"
    ],

    "B.Pharm": [
        "M.Sc / postgraduate specialization in a relevant life-science or healthcare field",
        "M.Sc Biotechnology / Microbiology / Life Sciences"
    ],

    "B.Ag": [
        "M.Sc Agriculture / Horticulture / Food Science specialization"
    ],

    "B.Des": [
        "M.Des / postgraduate specialization in Design or Media"
    ],

    "BA": [
        "MA in a relevant humanities or social-science specialization"
    ]
}

print("Eligibility mapping created successfully!")
print("Mapped education categories:", len(eligibility_mapping))

for education, pathways in eligibility_mapping.items():
    print(f"\n{education}:")
    for pathway in pathways:
        print("-", pathway)
```

```python
# Validate eligibility mapping against model career categories

model_careers = set(career_model.classes_)

mapping_results = []

for education, pathways in eligibility_mapping.items():
    for pathway in pathways:
        mapping_results.append({
            "Education Level": education,
            "Career Pathway": pathway,
            "Exists in Model": pathway in model_careers
        })

mapping_validation = pd.DataFrame(mapping_results)

display(mapping_validation)

print("\nTotal mapped pathways:", len(mapping_validation))
print(
    "Valid pathways:",
    mapping_validation["Exists in Model"].sum()
)
print(
    "Invalid pathways:",
    (~mapping_validation["Exists in Model"]).sum()
)
```

```python
def recommend_eligible_careers(student_profile, top_n=5):
    education = student_profile.iloc[0]["Education Level"]

    if education not in eligibility_mapping:
        return pd.DataFrame({
            "Message": [
                f"No preliminary eligibility rules configured for {education}."
            ]
        })

    # Generate model probabilities
    probabilities = career_model.predict_proba(student_profile)[0]

    # Build probability table
    predictions = pd.DataFrame({
        "Career": career_model.classes_,
        "Confidence (%)": probabilities * 100
    })

    # Filter to mapped pathways
    eligible_pathways = eligibility_mapping[education]

    recommendations = predictions[
        predictions["Career"].isin(eligible_pathways)
    ].copy()

    # Rank by model confidence
    recommendations = recommendations.sort_values(
        "Confidence (%)",
        ascending=False
    ).head(top_n).reset_index(drop=True)

    recommendations.index += 1
    recommendations.index.name = "Rank"

    return recommendations


print("Eligibility-aware recommendation function created!")
```

```python
btech_recommendations = recommend_eligible_careers(
    student_profile,
    top_n=5
)

print("EDUCATION LEVEL:", student_profile.iloc[0]["Education Level"])
print("\nELIGIBILITY-AWARE CAREER RECOMMENDATIONS")

display(btech_recommendations)
```

```python
def generate_recommendation_report(student_profile, recommendations):
    education = student_profile.iloc[0]["Education Level"]
    specialization = student_profile.iloc[0]["Specialization"]

    print("=" * 65)
    print("                 SKILLPATH AI")
    print("          PERSONALIZED CAREER REPORT")
    print("=" * 65)

    print(f"\nEducation Level : {education}")
    print(f"Specialization : {specialization}")

    print("\nRECOMMENDED CAREER PATHWAYS")
    print("-" * 65)

    if "Career" not in recommendations.columns:
        print("No preliminary pathway mapping is available.")
        return

    for rank, (_, row) in enumerate(
        recommendations.iterrows(), start=1
    ):
        print(f"\n{rank}. {row['Career']}")
        print(f"   Model score: {row['Confidence (%)']:.2f}%")

    print("\n" + "-" * 65)
    print("IMPORTANT:")
    print("These are preliminary, model-assisted suggestions.")
    print("Model scores are not admission eligibility probabilities.")
    print("Verify programme prerequisites and admission rules")
    print("with the relevant institution before applying.")
    print("=" * 65)


generate_recommendation_report(
    student_profile,
    btech_recommendations
)
```

```python
def recommend_eligible_careers(student_profile, top_n=5):
    education = student_profile.iloc[0]["Education Level"]
    specialization = student_profile.iloc[0]["Specialization"]

    # Check exact education-specialization mapping first
    profile_key = (education, specialization)

    if profile_key in specialization_mapping:
        eligible_pathways = specialization_mapping[profile_key]
        mapping_type = "Education + Specialization"

    elif education in eligibility_mapping:
        eligible_pathways = eligibility_mapping[education]
        mapping_type = "Education only"

    else:
        return pd.DataFrame({
            "Message": [
                "No preliminary pathway mapping is configured for this profile."
            ]
        })

    # Generate model probabilities
    probabilities = career_model.predict_proba(student_profile)[0]

    predictions = pd.DataFrame({
        "Career": career_model.classes_,
        "Confidence (%)": probabilities * 100
    })

    # Apply pathway filter
    recommendations = predictions[
        predictions["Career"].isin(eligible_pathways)
    ].copy()

    recommendations = recommendations.sort_values(
        "Confidence (%)",
        ascending=False
    ).head(top_n).reset_index(drop=True)

    recommendations.index += 1
    recommendations.index.name = "Rank"

    recommendations["Mapping Basis"] = mapping_type

    return recommendations


print("Updated specialization-aware recommendation function created!")
```

```python
class12_mapping = {
    ("Class 12", "MPC"): [
        "MPC → Intermediate (MPC) → B.Tech Computer Science pathway",
        "MPC → Intermediate (MPC) → B.Tech Mechanical/Electrical/Civil pathway"
    ],

    ("Class 12", "BiPC"): [
        "BiPC → Intermediate (BiPC) → Healthcare-related pathway",
        "BiPC → Intermediate (BiPC) → B.Sc Life Sciences pathway"
    ],

    ("Class 12", "CEC"): [
        "CEC / MEC → Intermediate → B.Com / Finance pathway"
    ],

    ("Class 12", "MEC"): [
        "CEC / MEC → Intermediate → B.Com / Finance pathway"
    ],

    ("Class 12", "HEC / Humanities"): [
        "HEC / Humanities → Intermediate → BA pathway"
    ],

    ("Class 12", "Arts/Humanities"): [
        "Arts / Humanities or a relevant design-oriented pathway"
    ]
}

print("Class 12 pathway mappings created!")
print("Configured stream combinations:", len(class12_mapping))
```

```python
def recommend_eligible_careers(student_profile, top_n=5):
    education = student_profile.iloc[0]["Education Level"]
    specialization = student_profile.iloc[0]["Specialization"]

    profile_key = (education, specialization)

    # Class 12 stream-based mapping
    if profile_key in class12_mapping:
        eligible_pathways = class12_mapping[profile_key]
        mapping_type = "Class 12 + Stream"

    # Degree-level specialization mapping
    elif profile_key in specialization_mapping:
        eligible_pathways = specialization_mapping[profile_key]
        mapping_type = "Education + Specialization"

    # Education-level fallback
    elif education in eligibility_mapping:
        eligible_pathways = eligibility_mapping[education]
        mapping_type = "Education only"

    else:
        return pd.DataFrame({
            "Message": [
                "No preliminary pathway mapping is configured for this profile."
            ]
        })

    probabilities = career_model.predict_proba(student_profile)[0]

    predictions = pd.DataFrame({
        "Career": career_model.classes_,
        "Confidence (%)": probabilities * 100
    })

    recommendations = predictions[
        predictions["Career"].isin(eligible_pathways)
    ].copy()

    recommendations = recommendations.sort_values(
        "Confidence (%)",
        ascending=False
    ).head(top_n).reset_index(drop=True)

    recommendations.index += 1
    recommendations.index.name = "Rank"

    recommendations["Mapping Basis"] = mapping_type

    return recommendations


print("Class 12 and degree-level recommendation function updated!")
```
